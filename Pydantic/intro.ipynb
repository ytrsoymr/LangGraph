# 🐍 Pydantic Basics

Pydantic is a Python library for data validation and settings management using Python type annotations.  
It enforces type hints at runtime and provides user-friendly errors when data is invalid.

This notebook covers:

- Creating Pydantic models
- Validating data types
- Using default values and optional fields
- Model configuration


```python
from pydantic import BaseModel
```

```python
class Person(BaseModel):
    name:str
    age:int
    city:str
    
person=Person(name="YTR",age=23,city="Hyderabad")
print(person)
```

```python
person1=Person(name="YTR",age=23,city=234)
```

## ❌ Invalid Data Example

Pydantic raises a `ValidationError` if the data does not match the specified types.


## 2. Model with optional Fields

## 🧩 Default Values and Optional Fields

Use `Optional` and default values for flexible model definitions.


```python
from typing import Optional
class Employee(BaseModel):
    id:int
    name:str
    dep:str
    sal:Optional[float]=None
    is_active:Optional[bool]=True
    
```

```python
Employee1=Employee(id=1,name="john",dep="IT")
print(Employee1)
```

```python
Employee2=Employee(id=1,name="john",dep="IT",sal=7000)
print(Employee2)
```

## 🧱 Nested Models

Pydantic supports nesting models within other models.


```python
class Address(BaseModel):
    city: str
    zip_code: str

class Customer(BaseModel):
    name: str
    address: Address

cust = Customer(name="John", address={"city": "New York", "zip_code": "10001"})
print(cust)
```

## ⚙️ Model Configuration with `Config`

You can customize how models behave using the `Config` class.


```python
class Person(BaseModel):
    name: str
    age: int

    class Config:
        anystr_strip_whitespace = True
        validate_assignment = True

person = Person(name=" Alice ", age=30)
print(person)
```
